**Analysis of failure modes**

```python
failure_columns = ["TWF", "HDF", "PWF", "OSF", "RNF"]
print(df[failure_columns].sum())
```

1. HDF has the highest amount of failure.
2. RNF has the lowest amount of failure.
3. The sum of failure_columns compared to the sum of columns "Machine Failure" because many of these columns may have failed at the same row.

***Failure Rate By Product Type***

```python
df.groupby("Type")["Machine failure"].agg(
    ['count', 'sum', 'mean']
)
```

```python
FailureRate = df.groupby("Type")["Machine failure"].mean()
print(FailureRate*100)
```
